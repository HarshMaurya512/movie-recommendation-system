```python
features = [
    "Genre",
    "Release_Year",
    "Rating",
    "Number_of_Ratings",
    "Runtime_Minutes",
    "Language",
    "Age_Group"
]

movie_features = df[features]

movie_features.head()
```

```python
# Step 4: Feature Preprocessing

X = movie_features.copy()

print("Preprocessing started...")
print(X.head())
```

```python
print("Missing values:")
print(X.isnull().sum())
```

```python
numeric_features = [
    "Release_Year",
    "Rating",
    "Number_of_Ratings",
    "Runtime_Minutes"
]

for col in numeric_features:
    X[col] = X[col].fillna(X[col].median())
```

```python
categorical_features = [
    "Genre",
    "Language",
    "Age_Group"
]

for col in categorical_features:
    X[col] = X[col].fillna(X[col].mode()[0])
```

```python
for col in categorical_features:
    X[col] = X[col].astype(str).str.strip()
```

```python
print(X.dtypes)
```

```python
print("Numerical Features:")
print(numeric_features)

print("\nCategorical Features:")
print(categorical_features)
```

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", StandardScaler(), numeric_features),
        ("categorical", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)
```

```python
# Step 5: Convert Features into Numerical Form

X_processed = preprocessor.fit_transform(X)

print("Features converted to numerical form successfully!")
```

```python
# Step 6: Create Feature Matrix

feature_matrix = X_processed

print("Feature matrix created successfully!")
print("Shape:", feature_matrix.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# Step 7: Calculate Cosine Similarity

similarity_matrix = cosine_similarity(feature_matrix)

print("Cosine similarity calculated successfully!")
print("Similarity matrix shape:", similarity_matrix.shape)
```

```python
# ==========================================
# STEP 8: BUILD RECOMMENDATION FUNCTION
# ==========================================

def recommend_movies(movie_title, num_recommendations=5):

    # Find the movie
    movie_indices = df.index[
        df["Movie_Title"].str.lower() == movie_title.lower()
    ].tolist()

    # Check if movie exists
    if not movie_indices:
        print("Movie not found!")
        return

    # Get the first matching movie index
    movie_index = movie_indices[0]

    # Get similarity scores for the selected movie
    similarity_scores = similarity_matrix[movie_index]

    # Sort movies from highest similarity to lowest
    similar_movie_indices = np.argsort(similarity_scores)[::-1]

    # Remove the selected movie itself
    similar_movie_indices = [
        i for i in similar_movie_indices
        if i != movie_index
    ]

    # Select top recommendations
    top_indices = similar_movie_indices[:num_recommendations]

    # Display recommendations
    print(f"\nMovies similar to: {df.iloc[movie_index]['Movie_Title']}")
    print("-" * 50)

    for rank, index in enumerate(top_indices, start=1):

        movie_name = df.iloc[index]["Movie_Title"]
        score = similarity_scores[index]

        print(f"{rank}. {movie_name} (Similarity: {score:.3f})")
```

```python
def get_user_history(user_id):
    
    history = df[df["User_ID"] == user_id].copy()
    
    if history.empty:
        print("User not found!")
        return
    
    print(f"Previously watched by User {user_id}:")
    print("-" * 50)
    
    for _, row in history.iterrows():
        print(
            f"{row['Movie_Title']} "
            f"| Genre: {row['Genre']} "
            f"| Rating: {row['Rating']}"
        )
```

```python
def recommend_for_user(user_id, num_recommendations=5):

    # Get user's history
    user_history = df[df["User_ID"] == user_id]

    if user_history.empty:
        print("User not found!")
        return

    # Get indices of movies already watched
    watched_indices = user_history.index.tolist()

    # Calculate recommendation scores
    recommendation_scores = similarity_matrix[watched_indices].mean(axis=0)

    # Remove movies already watched
    recommendation_scores[watched_indices] = -1

    # Sort by highest similarity
    recommended_indices = np.argsort(
        recommendation_scores
    )[::-1]

    # Get top recommendations
    recommended_indices = recommended_indices[
        :num_recommendations
    ]

    print(f"\nRecommendations for User {user_id}")
    print("-" * 50)

    for rank, index in enumerate(
        recommended_indices,
        start=1
    ):

        movie_name = df.iloc[index]["Movie_Title"]
        score = recommendation_scores[index]

        print(
            f"{rank}. {movie_name} "
            f"(Similarity: {score:.3f})"
        )
```

```python
recommend_for_user(157, 5)
```
